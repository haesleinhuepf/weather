# Downloading weather data from Huggingface

This notebook downloads weather data in [parquet format]() from Huggingface:
* [haesleinhuepf/dwd_temperature_july2026](https://huggingface.co/datasets/haesleinhuepf/dwd_temperature_july2026)
* [haesleinhuepf/dwd_precipitation_july2026](https://huggingface.co/datasets/haesleinhuepf/dwd_precipitation_july2026)

The data originated from Deutscher Wetterdienst (DWD) Open Data portal. 
Source: https://opendata.dwd.de/, licensed CC-BY 4.0 Read more: https://www.dwd.de/EN/service/legal_notice/legal_notice_node.html

```python
from datasets import load_dataset

ds = load_dataset("haesleinhuepf/dwd_temperature_july2026")

# Save one split as Parquet
ds["train"].to_parquet("temp_recent.parquet")
```

```python
ds = load_dataset("haesleinhuepf/dwd_precipitation_july2026")

# Save one split as Parquet
ds["train"].to_parquet("rain_recent.parquet")
```
