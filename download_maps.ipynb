# Maps of Germany

In this notebook we download and save GeoJson data of outlines of Germany and its federal states from [Amazons Registry of Open Data](https://registry.opendata.aws/naturalearth/). This data will later be used to overlay it with weather measurements.

```python
import geopandas as gpd
import matplotlib.pyplot as plt

url = "https://naturalearth.s3.amazonaws.com/10m_cultural/ne_10m_admin_0_countries.zip"

world = gpd.read_file(url)
germany = world[world["ADMIN"] == "Germany"]

fig, ax = plt.subplots(figsize=(6, 8))
germany.boundary.plot(ax=ax, color="black", linewidth=1)
ax.set_axis_off()
plt.show()
```

```python
import geopandas as gpd
import matplotlib.pyplot as plt

countries_url = "https://naturalearth.s3.amazonaws.com/10m_cultural/ne_10m_admin_0_countries.zip"
states_url = "https://naturalearth.s3.amazonaws.com/10m_cultural/ne_10m_admin_1_states_provinces.zip"

world = gpd.read_file(countries_url)
states = gpd.read_file(states_url)

germany = world[world["ADMIN"] == "Germany"]
states = states[states["admin"] == "Germany"]

fig, ax = plt.subplots(figsize=(6, 8))

germany.boundary.plot(ax=ax, color="black", linewidth=1.4)
states.boundary.plot(ax=ax, color="black", linewidth=0.6)

ax.set_axis_off()
plt.show()
```

```python
states.to_file("germany_federal_states.geojson", driver="GeoJSON")
germany.to_file("germany_outline.geojson", driver="GeoJSON")
```
