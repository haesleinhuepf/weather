# Uploading data to huggingface

This notebook demonstrates how the DWD data was uploaded / republished on Huggingface. The datasets can be found online:
* [dwd_temperature_july2026](https://huggingface.co/datasets/haesleinhuepf/dwd_temperature_july2026)
* [haesleinhuepf/dwd_precipitation_july2026](https://huggingface.co/datasets/haesleinhuepf/haesleinhuepf/dwd_precipitation_july2026)

```python
from huggingface_hub import HfApi, create_repo
```

```python
repo_id = "haesleinhuepf/dwd_temperature_july2026"
local_parquet = "temp_recent.parquet"

# Create the dataset repo if it does not exist
create_repo(
    repo_id=repo_id,
    repo_type="dataset",
    private=False,   # set True if needed
    exist_ok=True,
)

api = HfApi()

api.upload_file(
    path_or_fileobj=local_parquet,
    path_in_repo="temp_recent.parquet",
    repo_id=repo_id,
    repo_type="dataset",
    commit_message="Upload parquet dataset",
)
```

```python
repo_id = "haesleinhuepf/dwd_precipitation_july2026"
local_parquet = "rain_recent.parquet"

# Create the dataset repo if it does not exist
create_repo(
    repo_id=repo_id,
    repo_type="dataset",
    private=False,   # set True if needed
    exist_ok=True,
)

api = HfApi()

api.upload_file(
    path_or_fileobj=local_parquet,
    path_in_repo="rain_recent.parquet",
    repo_id=repo_id,
    repo_type="dataset",
    commit_message="Upload parquet dataset",
)
```
